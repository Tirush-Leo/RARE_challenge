```python
from sklearn.model_selection import train_test_split

# Load the simplified CSV file
destination_folder = "/content/drive/MyDrive/RARE25/Dataset/extracted"
new_output_csv_path = os.path.join(destination_folder, 'image_labels_simplified.csv')
df_simplified = pd.read_csv(new_output_csv_path)

# Split the data into training and testing sets (80% train, 20% test)
df_train, df_test = train_test_split(df_simplified, test_size=0.2, random_state=42, stratify=df_simplified['label'])

# Further split the training set into training and validation sets (80% train, 20% validation of the original train set)
df_train, df_val = train_test_split(df_train, test_size=0.25, random_state=42, stratify=df_train['label']) # 0.25 * 0.8 = 0.2 of original

# Print the shapes of the resulting DataFrames
print("Shape of training set:", df_train.shape)
print("Shape of validation set:", df_val.shape)
print("Shape of testing set:", df_test.shape)
```

## Model training

### Subtask:
Train a suitable classification model (e.g., a CNN with transfer learning) on the training data (`df_train`).


**Reasoning**:
Train a CNN model using transfer learning.



```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import VGG16
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Model
import numpy as np
from PIL import Image

# Define image dimensions and number of classes
img_height, img_width = 128, 128  # Adjust based on typical image size or desired input size
num_classes = 2  # Assuming binary classification based on the label column

# Create data generators
train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1./255)

# Create generators from DataFrame
train_generator = train_datagen.flow_from_dataframe(
    dataframe=df_train,
    directory="/content/drive/MyDrive/RARE25/Dataset/extracted", # Directory containing the images
    x_col="image_path",
    y_col="label",
    target_size=(img_height, img_width),
    batch_size=32,
    class_mode="raw" # Use "raw" for numerical labels
)

validation_generator = val_datagen.flow_from_dataframe(
    dataframe=df_val,
    directory="/content/drive/MyDrive/RARE25/Dataset/extracted", # Directory containing the images
    x_col="image_path",
    y_col="label",
    target_size=(img_height, img_width),
    batch_size=32,
    class_mode="raw" # Use "raw" for numerical labels
)

# Load pre-trained VGG16 model
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(img_height, img_width, 3))

# Add new layers for classification
x = base_model.output
x = Flatten()(x)
x = Dense(256, activation='relu')(x)
predictions = Dense(num_classes, activation='softmax')(x) # Use softmax for multi-class, sigmoid for binary

# Create the final model
model = Model(inputs=base_model.input, outputs=predictions)

# Freeze the layers of the pre-trained model
for layer in base_model.layers:
    layer.trainable = False

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy']) # Use sparse_categorical_crossentropy for integer labels

# Train the model
history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // train_generator.batch_size,
    epochs=10, # Adjust number of epochs
    validation_data=validation_generator,
    validation_steps=validation_generator.samples // validation_generator.batch_size
)
```

## Model evaluation

### Subtask:
Evaluate the trained model on the validation set, focusing on metrics like precision, recall, and F1-score. Specifically, analyze the precision at the desired recall level (90%).


**Reasoning**:
Evaluate the trained model on the validation set and calculate precision at 90% recall.



```python
import numpy as np
from sklearn.metrics import classification_report, precision_recall_curve

# Get true labels and predicted probabilities for the validation set
# We need to reset the generator to ensure we get predictions for the entire validation set
validation_generator.reset()
y_true = validation_generator.classes
y_pred_probs = model.predict(validation_generator)

# Calculate predicted class labels
y_pred = np.argmax(y_pred_probs, axis=1)

# Calculate and print the classification report
print("Classification Report on Validation Set:")
print(classification_report(y_true, y_pred))

# Calculate precision and recall for various thresholds
precision, recall, thresholds = precision_recall_curve(y_true, y_pred_probs[:, 1]) # Assuming class 1 is the positive class

# Find the precision at approximately 90% recall
# Iterate through the recall values and find the precision where recall is close to 0.9
precision_at_90_recall = None
for i in range(len(recall)):
    if recall[i] >= 0.9:
        precision_at_90_recall = precision[i]
        break

# Print the precision at 90% recall
print(f"Precision at approximately 90% Recall: {precision_at_90_recall}")
```
